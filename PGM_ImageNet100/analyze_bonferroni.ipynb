```python
import sqlite3
import json
import math
import numpy as np
from scipy.stats import ttest_1samp
```

```python
DBS = {
    "Run1": "/root/autodl-tmp/runs_main/pgm_main.db",
    "Run2": "/root/autodl-tmp/runs_main2/pgm_main.db",
    "Run3": "/root/autodl-tmp/runs_main3/pgm_main.db",
}

ALPHA = 0.1
M = 6
ALPHA_BONF = ALPHA / M
```

```python
def one_sided_pvalue(x):
    t, p_two = ttest_1samp(x, popmean=0.0)
    return p_two / 2 if t > 0 else 1.0

for run, db in DBS.items():
    conn = sqlite3.connect(db)
    rows = conn.execute("""
        SELECT iter, accepted, lb, mu_hat
        FROM proposals
        ORDER BY iter
    """).fetchall()
    conn.close()

    print(f"\n=== {run} ===")
    print("SGM accepts:", [r[0] for r in rows if r[1] == 1])
    print("NOTE: proposals table stores mu/lb, not raw paired deltas.")
    print("For exact Bonferroni t-test, extract paired deltas from logs or runs table.")
    print("Available proposal summary:")
    for it, acc, lb, mu in rows:
        print(f"iter {it}: SGM={acc}, mu={mu:.4f}, lb={lb:.4f}")
```
